# 3. Data Structures

In biomedical research, we rarely work with single values. We need structures to organize multiple data points.

    Lists: Ordered, mutable (changeable) collections of items. Great for storing a series of measurements or gene names.

    Dictionaries: Key-value pairs. Perfect for mapping identifiers to data, like mapping a patient ID to their clinical records.

    Tuples: Ordered, immutable (unchangeable) collections. Useful for fixed coordinates or data that shouldn't be altered.

```python
# LIST: A sequence of white blood cell counts over 4 days
wbc_counts = [4.5, 5.2, 7.8, 6.1]
wbc_counts.append(5.9) # Adding a new measurement for day 5
print("WBC Counts:", wbc_counts)

# DICTIONARY: Patient demographics
patient_record = {
    "age": 45,
    "blood_type": "O+",
    "allergies": ["Penicillin", "Peanuts"]
}
print("Patient's Blood Type:", patient_record["blood_type"])

# TUPLE: Genomic coordinates (Chromosome, Start, End, strand)
brca1_locus = ("chr17", 43044295, 43170327, "complement")
print("BRCA1 Chromosome:", brca1_locus[0], brca1_locus[3])
```

```python
# List of Dictionaries
cohort = [
    {"id": "P001", "disease": "Asthma", "age": 34},
    {"id": "P002", "disease": "Diabetes", "age": 52}
]
print("Second patient's disease:", cohort[1]["disease"])

```

#4. Functions

A function is a reusable block of code designed to perform a specific task. In data science, you will write functions to clean data, calculate statistics, or normalize expression values. They must be defined using the def keyword, followed by the function name, parentheses containing optional parameters, and a colon. The body of the function must be indented (typically 4 spaces or one tab).

```python
# Define a function to convert Celsius to Fahrenheit for clinical reporting
def celsius_to_fahrenheit(temperature_celsius):
    """Converts a Celsius temperature to Fahrenheit."""
    fahrenheit = (temperature_celsius * 9/5) + 32
    return fahrenheit
# Using the function for reporting the temperature in Farenheit for the patient with ID 1045 (38.5 Celsius)
temperature_celsius = 38.5
f_temp = celsius_to_fahrenheit(temperature_celsius)
print(f"Temperature in Fahrenheit: {f_temp}")
```
